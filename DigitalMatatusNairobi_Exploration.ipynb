```python
import urllib.request, zipfile, pandas as pd

url = "https://files.mobilitydatabase.org/mdb-1815/mdb-1815-202406071505/mdb-1815-202406071505.zip"
urllib.request.urlretrieve(url, "mdb-1815.zip")

with zipfile.ZipFile("mdb-1815.zip") as z:
    z.extractall("mdb-1815")

df = pd.read_csv("mdb-1815/stop_times.txt")
df.head()
```

```python
import pandas as pd

df = pd.read_csv('mdb-1815/stop_times.txt')
df.head()
```

a. How many rows and columns are contained in the dataset?

```python
# The `stop_times.txt` table contains 7,533 rows and 5 columns.

df.shape
```

b. What datatypes are contained in the dataset?

```python
# The `stop_sequence` column is stored as a whole number using the integer datatype. The remaining four columns—`trip_id`, `arrival_time`, `departure_time`, and
# `stop_id`—are stored as text (`object` in pandas). This means that the two time columns are text strings rather than actual time or date values.

df.dtypes
```

c. Is the dataset complete? i.e. no missing values?

```python
# All five columns have a missing-value count of zero, so every row in this table has a value in each column. This result applies only to `stop_times.txt`
# and does not prove that every table in the full GTFS feed is complete.

df.isnull().sum()
```
